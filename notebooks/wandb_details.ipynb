```python
import wandb
from collections import OrderedDict

def flatten_dict(d, parent_key='', sep='/'):
    """Flatten nested dictionary."""
    items = []
    for k, v in d.items():
        new_key = f"{parent_key}{sep}{k}" if parent_key else k
        if isinstance(v, dict):
            items.extend(flatten_dict(v, new_key, sep=sep).items())
        else:
            items.append((new_key, v))
    return dict(items)

def compare_configs(prev_config, curr_config):
    """Compare two configs and return differences."""
    prev_flat = flatten_dict(prev_config)
    curr_flat = flatten_dict(curr_config)
    
    changes = {
        'added': {},
        'removed': {},
        'modified': {}
    }
    
    # Find added and modified keys
    for key in curr_flat:
        if key not in prev_flat:
            changes['added'][key] = curr_flat[key]
        elif prev_flat[key] != curr_flat[key]:
            changes['modified'][key] = (prev_flat[key], curr_flat[key])
    
    # Find removed keys
    for key in prev_flat:
        if key not in curr_flat:
            changes['removed'][key] = prev_flat[key]
    
    return changes

# Initialize API
api = wandb.Api()
runs = api.runs("wandbleo/har-imu-training", per_page=10)

prev_config = None
prev_run_name = None

for run in runs:
    print(f"\n{'='*60}")
    print(f"Run: {run.name}")
    print(f"Created: {run.created_at}")
    
    # Get current config
    curr_config = dict(run.config)
    
    # Compare with previous run
    if prev_config is not None:
        changes = compare_configs(prev_config, curr_config)
        
        has_changes = any(changes.values())
        if has_changes:
            print(f"\n🔍 CHANGES from previous run ({prev_run_name}):")
            
            if changes['added']:
                print("\n  ➕ ADDED:")
                for key, value in changes['added'].items():
                    print(f"    {key}: {value}")
            
            if changes['removed']:
                print("\n  ➖ REMOVED:")
                for key, value in changes['removed'].items():
                    print(f"    {key}: {value}")
            
            if changes['modified']:
                print("\n  🔄 MODIFIED:")
                for key, (old_val, new_val) in changes['modified'].items():
                    print(f"    {key}:")
                    print(f"      OLD: {old_val}")
                    print(f"      NEW: {new_val}")
        else:
            print("\n  ✓ No changes from previous run")
    
    # Print all hyperparameters (with highlighting if changed)
    print(f"\nHyperparameters:")
    if prev_config is not None:
        changes = compare_configs(prev_config, curr_config)
        all_changes = set(changes['added'].keys()) | set(changes['modified'].keys())
    else:
        all_changes = set()
    
    for key, value in sorted(curr_config.items()):
        if isinstance(value, dict):
            print(f"  {key}:")
            for k, v in sorted(value.items()):
                full_key = f"{key}/{k}"
                marker = "  🔄" if full_key in all_changes else "    "
                print(f"{marker} {k}: {v}")
        else:
            marker = "🔄" if key in all_changes else " "
            print(f"  {marker} {key}: {value}")
    
    # Update for next iteration
    prev_config = curr_config
    prev_run_name = run.name
```

## print paramaters from best model file

```python
def inspect_checkpoint(checkpoint_path, verbose=False):
    """
    Inspect a PyTorch checkpoint and print architecture/hyperparameter info.
    
    Args:
        checkpoint_path: Path to .pth checkpoint file
        verbose: If True, print all layer keys with shapes
    """
    import torch
    from pathlib import Path
    
    def format_size(size_bytes):
        for unit in ['B', 'KB', 'MB', 'GB']:
            if size_bytes < 1024.0:
                return f"{size_bytes:.2f} {unit}"
            size_bytes /= 1024.0
        return f"{size_bytes:.2f} TB"
    
    def infer_architecture(state_dict):
        arch = {}
        
        # LLE parameters
        if 'lle.conv_blocks.0.0.weight' in state_dict:
            first_conv = state_dict['lle.conv_blocks.0.0.weight']
            arch['lle_in_channels'] = first_conv.shape[1]
            arch['lle_cnn_filters'] = []
            
            conv_idx = 0
            while f'lle.conv_blocks.{conv_idx}.0.weight' in state_dict:
                dilation_idx = 0
                total_filters = 0
                while f'lle.conv_blocks.{conv_idx}.{dilation_idx}.weight' in state_dict:
                    total_filters += state_dict[f'lle.conv_blocks.{conv_idx}.{dilation_idx}.weight'].shape[0]
                    dilation_idx += 1
                arch['lle_cnn_filters'].append(total_filters)
                conv_idx += 1
            
            if 'lle.gru.weight_ih_l0' in state_dict:
                gru_hidden = state_dict['lle.gru.weight_ih_l0'].shape[0] // 3
                arch['lle_gru_hidden'] = gru_hidden
                
                gru_layers = 0
                while f'lle.gru.weight_ih_l{gru_layers}' in state_dict:
                    gru_layers += 1
                arch['lle_gru_layers'] = gru_layers
            
            if 'lle.fc.weight' in state_dict:
                arch['lle_embedding_dim'] = state_dict['lle.fc.weight'].shape[0]
        
        # HLA parameters
        if 'hla.cls_token' in state_dict:
            arch['hla_type'] = 'transformer'
            arch['hla_embedding_dim'] = state_dict['hla.cls_token'].shape[2]
            
            if 'hla.pos_embed' in state_dict:
                arch['hla_seq_len'] = state_dict['hla.pos_embed'].shape[0] - 1
            
            if 'hla.encoder.layers.0.self_attn.in_proj_weight' in state_dict:
                in_proj = state_dict['hla.encoder.layers.0.self_attn.in_proj_weight']
                d_model = in_proj.shape[1]
                arch['hla_hidden_dim'] = d_model
                
                num_layers = 0
                while f'hla.encoder.layers.{num_layers}.self_attn.in_proj_weight' in state_dict:
                    num_layers += 1
                arch['hla_num_layers'] = num_layers
                
                if 'hla.encoder.layers.0.linear1.weight' in state_dict:
                    linear1 = state_dict['hla.encoder.layers.0.linear1.weight']
                    arch['hla_dim_feedforward'] = linear1.shape[0]
                    
                    possible_nheads = [1, 2, 4, 8]
                    for nhead in possible_nheads:
                        if d_model % nhead == 0:
                            arch['hla_nhead'] = nhead
                            break
            
            if 'hla.head.weight' in state_dict:
                arch['hla_num_classes'] = state_dict['hla.head.weight'].shape[0]
        
        elif 'hla.gru.weight_ih_l0' in state_dict:
            arch['hla_type'] = 'gru'
            gru_hidden = state_dict['hla.gru.weight_ih_l0'].shape[0] // 3
            arch['hla_hidden_dim'] = gru_hidden
            
            gru_layers = 0
            while f'hla.gru.weight_ih_l{gru_layers}' in state_dict:
                gru_layers += 1
            arch['hla_num_layers'] = gru_layers
            
            if 'hla.head.weight' in state_dict:
                arch['hla_num_classes'] = state_dict['hla.head.weight'].shape[0]
        
        if 'action_head.weight' in state_dict:
            arch['action_num_classes'] = state_dict['action_head.weight'].shape[0]
        
        return arch
    
    def count_parameters(state_dict):
        total_params = sum(p.numel() for p in state_dict.values())
        lle_params = sum(p.numel() for k, p in state_dict.items() if k.startswith('lle.') or k.startswith('module.lle.'))
        hla_params = sum(p.numel() for k, p in state_dict.items() if k.startswith('hla.') or k.startswith('module.hla.'))
        action_params = sum(p.numel() for k, p in state_dict.items() if k.startswith('action_head.') or k.startswith('module.action_head.'))
        return {
            'total': total_params,
            'lle': lle_params,
            'hla': hla_params,
            'action_head': action_params
        }
    
    # Load checkpoint
    print("="*80)
    print(f"CHECKPOINT INSPECTION: {checkpoint_path}")
    print("="*80)
    
    try:
        checkpoint = torch.load(checkpoint_path, map_location='cpu')
    except Exception as e:
        print(f"❌ Error loading checkpoint: {e}")
        return
    
    # Check format
    print("\n📦 Checkpoint Format:")
    if isinstance(checkpoint, dict):
        if 'model_state_dict' in checkpoint:
            print("   Format: Dictionary with 'model_state_dict' key")
            state_dict = checkpoint['model_state_dict']
            print(f"   Additional keys: {[k for k in checkpoint.keys() if k != 'model_state_dict']}")
            if 'epoch' in checkpoint:
                print(f"   Epoch: {checkpoint['epoch']}")
            if 'val_loss' in checkpoint:
                print(f"   Val Loss: {checkpoint['val_loss']:.4f}")
            if 'best_score' in checkpoint:
                print(f"   Best Score: {checkpoint['best_score']:.4f}")
        elif 'state_dict' in checkpoint:
            print("   Format: Dictionary with 'state_dict' key")
            state_dict = checkpoint['state_dict']
        else:
            print("   Format: Dictionary (assumed to be state_dict)")
            state_dict = checkpoint
    else:
        print("   Format: Direct state_dict (OrderedDict)")
        state_dict = checkpoint
    
    # Check for DataParallel prefix and strip it
    has_module_prefix = any(k.startswith('module.') for k in state_dict.keys())
    if has_module_prefix:
        print("   ⚠️  Contains 'module.' prefix (saved with DataParallel)")
        print("   🔧 Stripping 'module.' prefix for analysis...")
        # Strip the prefix
        new_state_dict = {}
        for k, v in state_dict.items():
            if k.startswith('module.'):
                new_state_dict[k[7:]] = v  # Remove 'module.' (7 chars)
            else:
                new_state_dict[k] = v
        state_dict = new_state_dict
    else:
        print("   ✓ No 'module.' prefix")
    
    # Show sample keys
    print(f"\n   Sample keys (first 5):")
    for i, key in enumerate(list(state_dict.keys())[:5]):
        print(f"      - {key}")
    
    # Parameter counts
    print("\n📊 Parameter Counts:")
    param_counts = count_parameters(state_dict)
    print(f"   Total parameters: {param_counts['total']:,}")
    if param_counts['total'] > 0:
        print(f"   LLE: {param_counts['lle']:,} ({param_counts['lle']/param_counts['total']*100:.1f}%)")
        print(f"   HLA: {param_counts['hla']:,} ({param_counts['hla']/param_counts['total']*100:.1f}%)")
        print(f"   Action Head: {param_counts['action_head']:,} ({param_counts['action_head']/param_counts['total']*100:.1f}%)")
    
    # Infer architecture
    print("\n🏗️  Inferred Architecture:")
    arch = infer_architecture(state_dict)
    
    if not arch:
        print("   ⚠️  Could not infer architecture. Showing all key prefixes:")
        prefixes = set()
        for key in state_dict.keys():
            prefix = key.split('.')[0]
            prefixes.add(prefix)
        print(f"   Found prefixes: {sorted(prefixes)}")
    else:
        if 'lle_in_channels' in arch:
            print(f"\n   LLE (Low-Level Encoder):")
            print(f"     Input channels: {arch['lle_in_channels']}")
            if 'lle_cnn_filters' in arch:
                print(f"     CNN filters: {arch['lle_cnn_filters']}")
            if 'lle_gru_hidden' in arch:
                print(f"     GRU hidden: {arch['lle_gru_hidden']}")
            if 'lle_gru_layers' in arch:
                print(f"     GRU layers: {arch['lle_gru_layers']}")
            if 'lle_embedding_dim' in arch:
                print(f"     Embedding dim: {arch['lle_embedding_dim']}")
        
        if 'hla_type' in arch:
            print(f"\n   HLA (High-Level Architecture):")
            print(f"     Type: {arch['hla_type']}")
            if 'hla_hidden_dim' in arch:
                print(f"     Hidden dim: {arch['hla_hidden_dim']}")
            if 'hla_num_layers' in arch:
                print(f"     Num layers: {arch['hla_num_layers']}")
            if 'hla_seq_len' in arch:
                print(f"     Seq len: {arch['hla_seq_len']}")
            if 'hla_nhead' in arch:
                print(f"     Attention heads: {arch['hla_nhead']}")
            if 'hla_dim_feedforward' in arch:
                print(f"     Feedforward dim: {arch['hla_dim_feedforward']}")
            if 'hla_num_classes' in arch:
                print(f"     Num classes: {arch['hla_num_classes']}")
        
        if 'action_num_classes' in arch:
            print(f"\n   Action Head:")
            print(f"     Num classes: {arch['action_num_classes']}")
    
    # File size
    file_size = Path(checkpoint_path).stat().st_size
    print(f"\n💾 File Size: {format_size(file_size)}")
    
    # Verbose: list all keys
    if verbose:
        print("\n🔑 All State Dict Keys:")
        for i, key in enumerate(sorted(state_dict.keys())):
            shape = state_dict[key].shape
            params = state_dict[key].numel()
            print(f"   {i+1:3d}. {key:60s} {str(shape):25s} ({params:,} params)")
    
    print("\n" + "="*80)
    return arch, param_counts
```
