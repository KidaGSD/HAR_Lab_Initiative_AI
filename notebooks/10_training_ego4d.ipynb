# Ego4D Model Training
This notebook trains the SSL model and SVDD detector on the processed Ego4D data.

```python
import os
from pathlib import Path
import sys
import torch
import matplotlib.pyplot as plt
import numpy as np

# Add scripts directory to path
sys.path.append(str(Path.cwd().parent / "scripts"))
sys.path.append(str(Path.cwd() / "scripts"))

from train_ego4d_model import train, Ego4DDataset, SSLModel, DeepSVDD
```

```python
# Configuration
class Args:
    target_uids_file = "../target_uids.csv"
    processed_dir = "../data/processed_ego4d"
    output_dir = "../models/ego4d_ssl"
    batch_size = 32
    epochs = 20
    lr = 1e-3
    device = "cuda" if torch.cuda.is_available() else "cpu"

args = Args()

# Ensure output directory exists
Path(args.output_dir).mkdir(parents=True, exist_ok=True)
```

```python
# Train the model
print(f"Training on device: {args.device}")
train(args)
```

```python
# Load SVDD model and visualize scores
svdd_state = torch.load(Path(args.output_dir) / "svdd_model.pth")
center = svdd_state['center']
R = svdd_state['R']

print(f"SVDD Center: {center[:5]}...")
print(f"SVDD Radius (R): {R}")
```
